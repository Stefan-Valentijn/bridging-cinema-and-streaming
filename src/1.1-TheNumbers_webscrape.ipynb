# The Numbers - web scraping

This first script web scrapes the release date, movie title, production budget, domestic gross and worldwide gross of movies as found on The Numbers website: https://www.the-numbers.com/movie/budgets/all

```python
# Import libraries
from webdriver_manager.chrome import ChromeDriverManager 
from selenium.webdriver.chrome.service import Service 
from selenium.webdriver.common.by import By
from selenium import webdriver 
from bs4 import BeautifulSoup 
import pandas as pd 
import requests
import time
import os

# Use Selenium to automatically open the website in a new browser
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service) 

# Open a new list
raw_thenumbers = []

# Loop through pages with offset
for offset in range(1, 6801, 100): # Note: at the time of web scraping there are no web pages after offset 6801

    # Define the url to find the data
    url = f"https://www.the-numbers.com/movie/budgets/all/{offset}"
    driver.get(url)
    time.sleep(1)
    
    # Click consent button for when the page is opened for the first time
    if offset == 1:
        button = driver.find_element(By.XPATH, "/html/body/div[2]/div[2]/div[2]/div[2]/div[2]/button[2]")
        button.click()
        time.sleep(2)

    # Start extraction
    print(f"Now web scraping: page with info from movie {offset} onwards")

    # Parse the website
    soup = BeautifulSoup(driver.page_source, 'html.parser')
    table = soup.find("table")
    allmovies = table.find_all("tr")
    
    for movie in allmovies[1:]: #skip header row
        # Obtain all cells
        cells = movie.find_all("td")
    
        # Divide all cells
        rank = cells[0].get_text()
        release_year = cells[1].get_text()
        title = cells[2].get_text()
        production_budget = cells[3].get_text()
        domestic_gross = cells[4].get_text()
        worldwide_gross = cells[5].get_text()
        timestamp = time.time() # add timestamp for research transparancy

        # Save the data
        raw_thenumbers.append({"rank": rank, 
                             "title": title,
                             "release_year": release_year,
                             "production_budget": production_budget, 
                             "domestic_gross": domestic_gross, 
                             "worldwide_gross": worldwide_gross,
                             "timestamp": timestamp})

    # Prevent to overload server so
    time.sleep(0.5)

# First create directory to be sure
os.makedirs('../data', exist_ok=True)

# Save to CSV
df = pd.DataFrame(raw_thenumbers)
df.to_csv('../data/raw/raw_thenumbers.csv', index = False, encoding ='utf-8')
print(f"\n Scraped {len(raw_thenumbers)} movies and saved it into 'raw_thenumbers.csv' at {time.ctime(time.time())}")
```
